```python
def func_sin_cos(angles, c = 0.):
    """This function computes sines and cosines

    parameters:
    angles: array of angles in radians
    c: constant to add to cosines
    """
    sines = np.sin(angles)
    cosines = np.cos(angles) + c  
    return sines, cosines

angles = np.arange(0, 2*np.pi, np.pi/2)
sins, coss = func_sin_cos(np.array([0, np.pi/2, np.pi]))

print(f"Angles: {angles}, Sines: {sins}, Cosines: {coss}")
print(f"Angles: {np.array2string(angles, precision=2, floatmode='fixed')}")
```
