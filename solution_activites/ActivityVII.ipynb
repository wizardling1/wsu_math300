### Q2: Lists and slicing

Create a list of at least six numbers. Using slicing, print (a) the first three elements, (b) the last three elements, (c) every other element, and (d) the list in reverse order.


```python
# TODO: your code for Q2 here
```

```python
my_list = [1, 2, 3, 4, 5]
print(f"my_list = {my_list}, type of my_list = {type(my_list)}") # Print the value of my_list and its type

sublist_a = my_list[:3] # This is a slicing operation: start index is the beginning of the list, end index is 3 (exclusive)
print(f"sublist_a = {sublist_a}, type of sublist_a = {type(sublist_a)}") # Print the value of sublist_a and its type

sublist_b = my_list[-3:] # This is a slicing operation: start index is -3, end index is the end of the list
print(f"sublist_b = {sublist_b}, type of sublist_b = {type(sublist_b)}") # Print the value of sublist_b and its type

sublist_c = my_list[::-1] # This is a slicing operation: start index is the end of the list, end index is the beginning of the list, step is -1
print(f"sublist_c = {sublist_c}, type of sublist_c = {type(sublist_c)}") # Print the value of sublist_c and its type
```

### Q3: Loops and accumulation

Using a `for` loop and `range()`, compute the sum of squares of the integers from 1 to 10 (i.e. $1^2 + 2^2 + \cdots + 10^2$). Then, using a `while` loop, find the smallest integer $n$ such that $1^2 + 2^2 + \cdots + n^2 > 100$, and print both $n$ and that sum.


```python
# TODO: your code for Q3 here
```

```python
sum = 0
for idx in range(1,11):
    sum += idx**2
print(f"The sum of squares from 1 to 10 is: {sum}") # Print the sum of squares from 1 to 10

n=1
sum=0
while sum <= 100:
    sum += n**2
    n += 1
print(f"The sum of squares from 1 to {n-1} is: {sum}") # Print the sum of squares from 1 to n-1
```
