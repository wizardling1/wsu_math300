### Q4: Building 3D Rotation Matrices

In 3D, rotating a vector by an angle $\theta$ (in radians) around one of the coordinate axes is done by multiplying it by a $3\times 3$ **rotation matrix**. Rotation about the $x$-axis leaves the $x$-coordinate unchanged and rotates the $y$-$z$ plane; rotation about the $y$-axis leaves $y$ unchanged; rotation about the $z$-axis leaves $z$ unchanged. The three matrices are:

$$
R_x(\theta) = \begin{bmatrix} 1 & 0 & 0 \\ 0 & \cos\theta & -\sin\theta \\ 0 & \sin\theta & \cos\theta \end{bmatrix}
\qquad
R_y(\theta) = \begin{bmatrix} \cos\theta & 0 & \sin\theta \\ 0 & 1 & 0 \\ -\sin\theta & 0 & \cos\theta \end{bmatrix}
\qquad
R_z(\theta) = \begin{bmatrix} \cos\theta & -\sin\theta & 0 \\ \sin\theta & \cos\theta & 0 \\ 0 & 0 & 1 \end{bmatrix}
$$

To rotate a vector $v$ (a 1D NumPy array with 3 entries) by angle $\theta$ around a given axis, compute the matrix-vector product $R(\theta)\, v$ with `@`.

**Your task:**
1. Write three functions, `Rx(theta)`, `Ry(theta)`, and `Rz(theta)`, each taking an angle `theta` in radians and returning the corresponding $3\times 3$ rotation matrix above as a NumPy array (use `np.cos` and `np.sin`).
2. Create the vector `v = np.array([1, 0, 0])`. Using your three functions, rotate `v` by `theta = np.pi / 2` around the $x$-axis, then separately around the $y$-axis, then separately around the $z$-axis, and print all three results. Think about why each result makes sense: `v` starts pointing along the $x$-axis, so which rotation(s) should leave it unchanged?
3. a) Rotate `v` by `theta = np.pi / 4` (45 degree) around the $z$-axis, then rotate that result by another `np.pi / 4` radians (45 degrees) around the $z$-axis (i.e., apply `Rz` twice in a row). b) Compare this to rotating the original `v` once by `np.pi / 2` radians (90 degrees) around the $z$-axis, using `np.allclose(arr1, arr2)` with `arr1` and `arr2` being the NumPy array of your final results from parts a) and b), and print whether they agree.

```python
# TODO: your code for Q4 here
```

```python
def mat_rx(theta):
    c =np.cos(theta)
    s =np.sin(theta)
    return np.array([[1, 0, 0], [0, c, -s], [0, s, c]]) # Return the 3x3 rotation matrix about the x-axis

def mat_ry(theta):
    c =np.cos(theta)
    s =np.sin(theta)
    return np.array([[c, 0, s], [0, 1, 0], [-s, 0, c]]) # Return the 3x3 rotation matrix about the y-axis

def mat_rz(theta):
    c =np.cos(theta)
    s =np.sin(theta)
    return np.array([[c, -s, 0], [s, c, 0], [0, 0, 1]]) # Return the 3x3 rotation matrix about the z-axis

v= np.array([1, 0, 0]) # A vector along the x-axis

rx = mat_rx(np.pi/2) # 90 degree rotation about the x-axis
ry = mat_ry(np.pi/2) # 90 degree rotation about the y-axis
rz = mat_rz(np.pi/2) # 90 degree rotation about the z-axis

print(f"Rotation about x-axis: {rx @ v}. This keeps the vecor intact.") # Apply the rotation matrix to the vector
print(f"Rotation about y-axis: {ry @ v}") # Apply the rotation matrix to the vector
print(f"Rotation about z-axis: {rz @ v}") # Apply the rotation matrix to the vector

rx_pi4 = mat_rz(np.pi/4) # 45 degree rotation about the z-axis
print(f"Rotation about z-axis by 45 degrees: {rx_pi4 @ v}")
twice_rotated = rx_pi4 @ (rx_pi4 @ v) # Apply the rotation matrix twice to the vector
print(f"Rotation about z-axis by another 45 degrees: {np.array2string(twice_rotated, suppress_small=True, precision=2)}") # Apply the rotation matrix twice to the vector

vz_pi2 = mat_rz(np.pi/2) @ v # 90 degree rotation about the z-axis
print(np.allclose(vz_pi2, twice_rotated)) # Check if the two methods give the same result

```
