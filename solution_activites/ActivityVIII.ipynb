### Q2: For loops

Using a single `for` loop over `range(1, 11)` (no `if` statements needed), build a list containing the cube ($n^3$) of every number from `1` to `10`. Print the resulting list, and then use the built-in `sum()` function to print the total of all the cubes.

```python
# TODO: your code for Q2 here
```

### Q4: Combining loops and conditionals

```python
# TODO: your code for Q4 here
```

```python
sum = 0
for idx in range(1,51):
    if (idx % 3 == 0) or (idx % 5 == 0):
         sum +=idx 

print(f"sum of multiples of 3 or 5 below 51 = {sum}")

n = 0
sqr = 0
while sqr <= 500:
    sqr = n**2
    n += 1
print(f"The smallest integer n such that n^2 > 500 is {n-1}, with n^2 = {sqr}")
```

### Q5: Functions, loops, and conditionals together

Write a function `is_prime(n)` that returns `True` if `n` is a prime number (an integer greater than 1 with no divisors other than 1 and itself) and `False` otherwise. Inside the function, use a `for` loop over the possible divisors and an `if` statement to check divisibility; you can assume `n` is a positive integer.

Then, using your function inside a `for` loop over `range(2, 50)`, build a list of all the prime numbers less than 50, and print the list.

```python
# TODO: your code for Q5 here
```

```python
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

for idx in range(2,50):
    if is_prime(idx):
        print(f"{idx} is prime.")
```
