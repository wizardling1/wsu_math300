### Exercise 5

Using a dictionary, store the names and scores of three students. Then write a loop that prints each student's name together with their letter grade (A: 90+, B: 80-89, C: 70-79, F: below 70).

```python
# Your code here
```

```python
scores = {"Amara": 92, "Ben": 78, "Cleo": 65}
for name, score in scores.items():
    if score >= 90:
        grade = "A"
    elif score >= 80:
        grade = "B"
    elif score >= 70:
        grade = "C"
    else:
        grade = "F"
    print(name, grade)
```
