# Useful iterable: `range()`

The `range(start, stop, step)` function generates a sequence of numbers without storing them all in memory at once. It's commonly used for looping a fixed number of times. Like slicing, `stop` is exclusive and `start`/`step` are optional (defaulting to `0` and `1`).

`range()` produces a special `range` object, not a list; wrap it in `list(...)` to see all its values as an actual list.

````python
range_ints = range(10) # This is a range object
print(range_ints, type(range_ints)) # Print the value of range_ints and its type

list_ints = list(range_ints) # Convert the range object to a list
print(list_ints, type(list_ints)) # Print the value of list_ints and its type

range2 = range(1, 10, 2) # This is a range object with start, stop, and step
print(range2, type(range2)) # Print the value of range2 and its type

list_range2 = list(range2) # Convert the range object to a list
print(list_range2, type(list_range2)) # Print the value of list_range2 and its type
````

# If statements

An `if` statement lets your code make decisions: a block of code only runs **if** a certain condition is `True`. You can chain further conditions with `elif` ("else if"), and provide a fallback with `else` that runs when none of the previous conditions were `True`.

In general form:

```
if condition1:
    # runs if condition1 is True
elif condition2:
    # runs if condition1 is False and condition2 is True
else:
    # runs if none of the above were True
```

Only **one** of the branches runs, whichever condition is met first; `elif` and `else` are both optional, and you can chain as many `elif`s as you like.

````python
temperature = 15 # Try changing this value

if temperature > 25:
    print("It's hot outside.")
elif temperature > 10:
    print("It's mild outside.")
else:
    print("It's cold outside.")
````

Conditions can be combined using the **logical operators** `and`, `or`, and `not`:

| Operator | Meaning |
|---|---|
| `and` | `True` only if both sides are `True` |
| `or` | `True` if at least one side is `True` |
| `not` | flips `True` to `False` and vice versa |

You can also check whether a value is present in a list with the `in` keyword, e.g. `3 in [1, 2, 3]` is `True`.

````python
x = 7
is_prime_candidate = True  # imagine this came from some earlier check

# 'and': the whole condition is True only if (x > 0) is True AND (x < 10) is True
if (x > 0) and (x < 10):
    print(f"{x} is in the interval (0, 10).")
else:
    print(f"{x} is not in the interval (0, 10).")

known_primes = [2, 3, 5, 7, 11, 13, 17, 19]
n = 7

# 'in': membership check against a list of known primes
if n in known_primes:
    print(f"{n} is one of the primes we've already found!")

is_even = (x % 2 == 0)

# 'or': the whole condition is True if EITHER is_even is True OR is_prime_candidate is True
if is_even or is_prime_candidate:
    print(f"{x} is even, or a prime candidate (or both)!")

# 'not': flips a True/False value to its opposite
if not is_even:
    print(f"{x} is not even.")
````

### Piecewise functions with `if` / `elif` / `else`

Chained `if` / `elif` / `else` branches are a natural way to implement a **piecewise function**: a function whose formula changes depending on which range the input falls into. For example:

$$f(x)=\begin{cases} x+2 & \text{if } 1 \le x \le 2 \\ x-2 & \text{if } 2 < x \le 3 \\ x^2 & \text{if } x > 3 \\ 2x & \text{otherwise} \end{cases}$$

Each `elif`/`else` branch below only runs if none of the conditions above it were `True`, so the order of the conditions matters here.

````python
x = 2.5 # Try changing this value to land in each of the four cases

if 1 <= x <= 2:      # first piece: 1 <= x <= 2
    result = x + 2
elif 2 < x <= 3:      # second piece: 2 < x <= 3 (we already know x > 2 here)
    result = x - 2
elif x > 3:           # third piece: x > 3
    result = x ** 2
else:                 # otherwise: x < 1
    result = 2 * x

print(f"f({x}) = {result}")
````

# Iterations

**Loops** let you repeat a block of code multiple times. Python has two main kinds: `for` loops (repeat once per item in a sequence) and `while` loops (repeat as long as a condition is true).

### For loops

A `for` loop iterates over a sequence (like a `range` or a list), running the indented block of code once for each item. The loop variable (`i` below) takes on each value in the sequence in turn.

In general form, a `for` loop always looks like this:

```
for loop_variable in some_iterable:
    # this indented block runs once per item in some_iterable
    # loop_variable holds the current item on each pass
```

`some_iterable` can be a `range`, a list, or anything else Python can step through one item at a time.

````python
for i in range(5): # This is a for loop
    x = i + 2
    print(f"i = {i}, x = {x}") # Print the value of i and x
````

You can also loop directly over the elements of a list, rather than over indices. Here `idx` takes each value from the list `[2, 3, 4, 6]` in order.

Note that reassigning the loop variable inside the loop body (`idx = idx - 1`) only changes that variable for the current iteration; it doesn't affect the original list or the next value the loop will use.

````python
for idx in [2, 3, 4, 6]: # This is a for loop over a list
    print(f"idx = {idx} before subtraction") # Print the value of idx
    idx = idx - 1
    print(f"idx = {idx} after subtraction") # Print the value of idx after subtraction
````

You can also build up a list while looping, by starting with an empty list and calling `.append()` inside the loop. This "accumulator" pattern is one of the most common uses of `for` loops.

````python
squares = [] # Start with an empty list
for i in range(1, 6): # Loop over 1, 2, 3, 4, 5
    squares.append(i ** 2) # Append i squared to the list
print(squares, type(squares)) # Print the final list and its type
````

Loops can also be **nested**: you can put a `for` loop inside another `for` loop. The inner loop runs all the way through for *every single* pass of the outer loop, so the indented body of the inner loop ends up running `len(outer) * len(inner)` times in total.

````python
count = 0 # Running count of total inner-loop passes
for i in range(3): # Outer loop: i = 0, 1, 2
    for j in range(2): # Inner loop: j = 0, 1 (runs fully for each i)
        count += 1
        print(f"i = {i}, j = {j}, count = {count}") # Print the current state
````

### `break` and `continue`

Two extra keywords let you change how a loop runs before it finishes on its own, and both work inside `for` loops and `while` loops:

- `break` immediately exits the loop entirely; any remaining iterations are skipped.
- `continue` skips just the rest of the **current** iteration; the loop then moves on to the next item as usual.


````python
numbers = [4, 7, 2, 9, 5, 1, 8] # The list we're searching through
target = 9 # The value we're looking for - defined above the loop

for n in numbers: # Loop over each number in the list
    print(f"Checking {n}...") # Show every number we look at, until we stop early
    if n == target:
        break # Found it! No need to keep checking the rest of the list, so stop the loop entirely

print(f"Stopped as soon as we found {target} - break skipped checking {numbers[numbers.index(target)+1:]}\n")

for i in range(10): # Loop over 0 through 9
    if i % 2 == 0:
        continue # Skip the rest of THIS iteration (the print below) whenever i is even
    print(f"i = {i}") # Only prints the odd numbers: 1, 3, 5, 7, 9

print("Loop ran through all 10 iterations, just skipping some of them")
````

# While loops

A `while` loop repeats its block of code **as long as** its condition remains `True`. Unlike a `for` loop, the number of iterations isn't fixed in advance; it depends on when the condition becomes `False`.

In general form:

```
while condition:
    # this indented block runs repeatedly, as long as condition is True
    # something in here should eventually make condition False
```

It's important that something inside the loop eventually makes the condition `False` (here, `x` is decreased by `1` each time), otherwise the loop would run forever!

````python
x = 10
while x > 0:
    x = x - 1
    print(f"x = {x}: ", "The condition is ", x > 0) # Print the value of x after subtraction
````

`while` loops are especially useful when you don't know in advance how many iterations you'll need, for example, when you're looking for the first value that satisfies some condition.

````python
total = 0 # Running total
n = 0 # Counter
while total <= 50: # Keep going until the total passes 50
    n = n + 1
    total = total + n
    print(f"n = {n}, total so far = {total}") # Print the value of n and the running total

print(f"The total first exceeded 50 once n = {n}, with total = {total}") # Print the final result
````

`if` statements are often combined with loops: the loop visits each item, and the `if` decides what to do with it. Below, we build a list of only the even numbers from `0` to `9` using the modulo operator `%` to test whether a number is divisible by `2`.

````python
evens = [] # Start with an empty list

for i in range(10):
    if i % 2 == 0: # i % 2 is 0 exactly when i is even
        evens.append(i)

print(evens, type(evens)) # Print the final list of even numbers
````

# Functions

A **function** is a reusable, named block of code that can take inputs (called **parameters** or **arguments**) and can send back an output with the `return` statement. Functions let you avoid repeating yourself: write the logic once, then call it as many times as you like with different inputs.

A function is defined with the `def` keyword, followed by its name, its parameters in parentheses, and a colon. The indented block underneath is the function's body.

```
def function_name(parameter1, parameter2):
    # this indented block is the function body
    # it can use parameter1 and parameter2
    return some_value  # sends some_value back to the caller
```

````python
def square(x): # Define a function called square, with one parameter x
    return x ** 2 # Send back x squared

result = square(5) # Call the function with x = 5
print(result, type(result)) # Print the value of result and its type

print(square(3)) # You can also call a function directly inside print()
print(square(3) + square(4)) # Function calls can be combined like any other value
````

Functions can take **multiple parameters**, and parameters can have **default values**: if the caller doesn't provide that argument, the default is used instead.

````python
def power(base, exponent=2):  # exponent has a default value of 2
    result = base ** exponent
    return result

print(power(5))                        # Uses the default exponent: 5**2 = 25
print(power(5, 3))                     # Overrides the default: 5**3 = 125
print(power(base=2, exponent=10))      # Arguments can also be passed by name: 2**10 = 1024
````

Functions can also contain loops and `if` statements in their body, just like any other code. This lets you package more complex logic behind a single, reusable name.

````python
def count_evens(numbers): # numbers is expected to be a list
    count = 0 # Running count of even numbers
    for n in numbers: # Loop over every element in the list
        if n % 2 == 0: # Check whether n is even
            count = count + 1
    return count

print(count_evens([1, 2, 3, 4, 5, 6])) # Should be 3: 2, 4, 6
print(count_evens([1, 3, 5])) # Should be 0: no even numbers
````

### Q5: Functions, loops, and conditionals together

Write a function `is_prime(n)` that returns `True` if `n` is a prime number (an integer greater than 1 with no divisors other than 1 and itself) and `False` otherwise. Inside the function, use a `for` loop over the possible divisors and an `if` statement to check divisibility; you can assume `n` is a positive integer.

Then, using your function inside a `for` loop over `range(2, 50)`, build a list of all the prime numbers less than 50, and print the list.

````python
def is_prime(n: int) -> int: 
    if n < 2: return False
    for i in range(2, n-1):
        if n % i == 0: return False
    return True

for n in range(2,50):
    print(f"is {n} prime? {is_prime(n)}")
````
