# If statement

if cond1:
   any code block runs when cond 1 is true
elif cond2:
   a block of code runs when cond2 is true
   ...
else:
    This block runs when none of above block runs




```python
x = 210

if x <20:
    print("x is less than 20")
else:
    print("x is greater than or equal to 20")
```

```python
x = 28

if (x>0) and (x %2 ==0):
    print(f"x={x} is positive and even")
elif (x>0) and (x %2 !=0):
    print(f"x={x} is positive and odd")
else:
    print(f"x={x} is not positive or not an integer")
```

```python
sum = 0
n=1
while sum < 100:
    sum += n # sum = sum + n
    n += 1 # n = n+1
    print(f"sum={sum}, n={n}")
    if n > 5:
        break
```
