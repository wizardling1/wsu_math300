# More List Commands

You already know `len()`, indexing, `.append()`, and slicing from Part I. Lists have several other useful methods for adding, removing, and reordering elements:

- `list1.insert(i, value)` inserts `value` at position `i`, shifting later elements to the right.
- `list1.remove(value)` removes the first occurrence of `value` from the list (by value, not by position).
- `list1.pop(i)` removes and returns the element at position `i` (default: the last element).
- `list1.extend(list2)` appends every element of `list2` to `list1`, one by one (unlike `.append(list2)`, which would add `list2` as a single nested element).
- `list1.count(value)` returns how many times `value` appears in the list.
- `list1.index(value)` returns the position of the first occurrence of `value`.

Suppose you're solving the equation $x^3 - 6x^2 + 11x - 6 = 0$ numerically, and you're keeping a running list of candidate roots as you find them.

```python
roots = [1, 2, 5] # Candidate roots found so far (5 is a mistake we'll fix)
print(roots, type(roots)) # Print the value of roots and its type

roots.insert(0, 0.5) # Insert 0.5 at the front, another candidate to check
print(roots) # Print the updated list

roots.remove(5) # Remove the incorrect candidate 5
print(roots) # Print the updated list

popped = roots.pop() # Remove and return the last element
print(popped, roots) # Print the popped value and the remaining list

popped2 = roots.pop(0) # Remove and return the element at position 0
print(popped2, roots) # Print the popped value and the remaining list

roots.extend([3, 4]) # A classmate found two more candidates: add them individually
print(roots) # Print the updated list

n = roots.count(3) # Count how many times 3 appears in our candidate list
print(n) # Print the count

idx = roots.index(4) # Find the position of the candidate 4
print(idx) # Print the index
```

Lists also have methods for reordering their elements in place:

- `list1.sort()` sorts the list in ascending order, modifying it in place (it returns `None`, not the sorted list).
- `list1.sort(reverse=True)` sorts in descending order instead.
- `list1.reverse()` reverses the order of the elements in place.
- The built-in `sorted(list1)` function, by contrast, returns a new sorted list and leaves the original list unchanged.

```python
exam_scores = [78, 92, 65, 88, 100, 71] # Scores from a math exam
print(exam_scores) # Print the original list

exam_scores.sort() # Sort the list in place, ascending order
print(exam_scores) # Print the sorted list

exam_scores.sort(reverse=True) # Sort the list in place, descending order
print(exam_scores) # Print the sorted list

exam_scores.reverse() # Reverse the current order of the list in place
print(exam_scores) # Print the reversed list

unsorted_scores = [78, 92, 65, 88, 100, 71]
ranked_scores = sorted(unsorted_scores) # Returns a NEW sorted list, doesn't modify unsorted_scores
print(ranked_scores, type(ranked_scores)) # Print the new sorted list
print(unsorted_scores) # unsorted_scores is unchanged
```

One subtlety worth knowing: when you write `list2 = list1`, you are not making a copy; `list2` becomes another name for the same list, so modifying one modifies the other. To make an independent copy, use `.copy()` (or the slice `list1[:]`).

```python
coefficients = [1, -6, 11, -6] # Coefficients of x^3 - 6x^2 + 11x - 6
coefficients_alias = coefficients # This is the SAME list as coefficients, not a copy
coefficients_alias.append(0) # Append a constant term of 0 to the alias
print(coefficients) # coefficients was also changed!

coefficients_copy = coefficients.copy() # coefficients_copy is an independent copy
coefficients_copy.append(99)
print(coefficients) # coefficients is unaffected
print(coefficients_copy) # only coefficients_copy has the new element
```

# Tuples

A tuple is an ordered collection of values, written with parentheses `(...)` and separated by commas. Tuples behave a lot like lists (you can index and slice them), but with one key difference: tuples are immutable, meaning once created, their contents cannot be changed. There's no `.append()`, `.insert()`, or `.remove()` for tuples.

Tuples are a natural way to represent a point in the plane, since the coordinates $(x, y)$ of a point shouldn't accidentally change once you've defined it.

```python
point = (3, 4) # This is a tuple, representing the point (3, 4)
print(point, type(point)) # Print the value of point and its type

x_coord = point[0] # Indexing works just like lists
print(x_coord, type(x_coord)) # Print the value of x_coord and its type

sub = point[0:1] # Slicing also works just like lists
print(sub, type(sub)) # Print the value of sub and its type

l = len(point) # len() works on tuples too
print(l, type(l)) # Print the length of point

try:
    point[0] = 10 # This will fail: tuples cannot be modified
except TypeError as e:
    print(f"Error: {e}") # Print the error message
```

A very common use of tuples is unpacking: assigning each element of a tuple to its own variable in a single line. You've actually already seen this idea in Part I with `x, y = 10, 20`, that's tuple unpacking too!

Note: to create a tuple with a single element, you need a trailing comma, e.g. `(5,)`. Without the comma, `(5)` is just the integer `5` in parentheses.

```python
point = (3, 4)
x_coord, y_coord = point # Unpack the tuple into two variables
print(f"x_coord = {x_coord}, y_coord = {y_coord}") # Print both values

single = (5,) # A one-element tuple needs a trailing comma
not_a_tuple = (5) # This is just the integer 5
print(single, type(single)) # Print the value of single and its type
print(not_a_tuple, type(not_a_tuple)) # Print the value of not_a_tuple and its type

# Functions can also return multiple values, packaged as a tuple
def midpoint(p1, p2): # p1 and p2 are (x, y) points
    mx = (p1[0] + p2[0]) / 2
    my = (p1[1] + p2[1]) / 2
    return mx, my # This packs mx and my into a tuple automatically

m = midpoint((0, 0), (6, 8)) # Find the midpoint between two points
print(m, type(m)) # Print the resulting tuple

mx, my = midpoint((0, 0), (6, 8)) # Unpack the returned tuple directly
print(f"midpoint = ({mx}, {my})") # Print both coordinates
```

# Sets

A set is an unordered collection of unique values, written with curly braces `{...}`. Unlike lists and tuples, sets automatically remove duplicates and don't keep track of order or position, so sets can't be indexed or sliced.

Sets are useful whenever you care about membership (is this value present?) or want to eliminate duplicates from a collection, such as the set of distinct roots of an equation.

Note: to create an empty set, you must use `set()`, not `{}` (which creates an empty dictionary, as you'll see below).

```python
roots = {2, -1, 2, 3, -1} # The equation gave repeated roots; a set keeps only distinct values
print(roots, type(roots)) # Print the value of roots and its type

l = len(roots) # len() works on sets too
print(l, type(l)) # Print the number of distinct roots

roots.add(0) # Add a newly discovered root
print(roots) # Print the updated set

roots.add(2) # Adding a value that's already present has no effect
print(roots) # Print the set: still no duplicate 2

roots.remove(-1) # Remove a root (raises an error if not present)
print(roots) # Print the updated set

roots.discard(100) # Like remove(), but does NOT raise an error if not present
print(roots) # Print the set: unchanged

is_root = 3 in roots # Membership check with 'in', very fast for sets
print(is_root, type(is_root)) # Print the value of is_root and its type

empty_set = set() # This is how you create an empty set
print(empty_set, type(empty_set)) # Print the value of empty_set and its type

sample_values = [2, 2, 3, 4, 4, 4, 5] # A sequence of values from repeated calculations
distinct_values = set(sample_values) # Convert to a set to find the distinct values
print(distinct_values, type(distinct_values)) # Print the distinct values
```

Sets support the mathematical set operations you might already know from a discussion of divisors or number theory:

| Operator / Method | Meaning |
|---|---|
| `A \| B` or `A.union(B)` | all elements in `A` or `B` (or both) |
| `A & B` or `A.intersection(B)` | elements in both `A` and `B` |
| `A - B` or `A.difference(B)` | elements in `A` but not in `B` |
| `A ^ B` or `A.symmetric_difference(B)` | elements in exactly one of `A`, `B` (not both) |

For example, let `A` be the set of divisors of `12` and `B` be the set of divisors of `18`. The intersection `A & B` is exactly the set of common divisors of 12 and 18, and its largest element is their greatest common divisor.

```python
A = {1, 2, 3, 4, 6, 12} # Divisors of 12
B = {1, 2, 3, 6, 9, 18} # Divisors of 18

union_set = A | B # All divisors of 12 or 18 (or both)
print(union_set) # Print the union

common_divisors = A & B # Divisors shared by both 12 and 18
print(common_divisors) # Print the intersection

gcd_12_18 = max(common_divisors) # The greatest common divisor is the largest common divisor
print(gcd_12_18) # Print the greatest common divisor

only_divides_12 = A - B # Divides 12 but not 18
print(only_divides_12) # Print the difference

sym_diff_set = A ^ B # Divides exactly one of the two numbers, not both
print(sym_diff_set) # Print the symmetric difference
```

# Dictionaries

A dictionary stores a collection of `key: value` pairs, written with curly braces `{key1: value1, key2: value2, ...}`. Instead of accessing elements by numeric position (like a list), you access them by their key, which can be a string, a number, or any other immutable type.

Dictionaries are a natural way to represent the coefficients of a polynomial: each power of $x$ is a key, and its coefficient is the value. Consider $f(x) = 3x^2 - 5x + 2$.

```python
coeffs = {"x^2": 3, "x^1": -5, "x^0": 2} # Coefficients of f(x) = 3x^2 - 5x + 2
print(coeffs, type(coeffs)) # Print the value of coeffs and its type

leading_coeff = coeffs["x^2"] # Access a value by its key
print(leading_coeff, type(leading_coeff)) # Print the value of leading_coeff and its type

l = len(coeffs) # len() returns the number of key:value pairs
print(l, type(l)) # Print the number of terms

coeffs["x^1"] = -4 # Update the coefficient of x to -4
print(coeffs) # Print the updated dictionary

coeffs["x^3"] = 1 # Adding a new key automatically creates the entry
print(coeffs) # Print the updated dictionary: now a cubic

has_x2 = "x^2" in coeffs # 'in' checks the KEYS of a dictionary, not the values
print(has_x2, type(has_x2)) # Print the value of has_x2 and its type

del coeffs["x^3"] # Remove a key:value pair with 'del'
print(coeffs) # Print the updated dictionary

removed_value = coeffs.pop("x^0") # Remove a key and return its value
print(removed_value, coeffs) # Print the removed value and the remaining dictionary
```

Using `coeffs["key"]` for a key that doesn't exist raises an error. The `.get(key, default)` method is a safer alternative: it returns `default` (or `None` if you don't specify one) instead of crashing, for example if $f(x)$ has no constant term.

To loop over a dictionary, you can use `.keys()`, `.values()`, or `.items()` (which gives you `key, value` pairs together, ready for unpacking in a `for` loop).

```python
coeffs = {"x^2": 3, "x^1": -5} # f(x) = 3x^2 - 5x, no constant term

leading = coeffs.get("x^2", 0) # Key exists, returns its value
print(leading) # Print the value

constant = coeffs.get("x^0", 0) # Key doesn't exist, returns the default 0 instead
print(constant) # Print the default value

print("Powers of x:") # Loop over just the keys
for power in coeffs.keys():
    print(f"  {power}")

print("Coefficients:") # Loop over just the values
for value in coeffs.values():
    print(f"  {value}")

print("Terms:") # Loop over key, value pairs together
for power, value in coeffs.items():
    print(f"  {value} * {power}")
```

Dictionary values can be any type, including lists, tuples, or even other dictionaries, which lets you represent more complex, nested data, such as a gradebook mapping each student's name to their list of quiz scores.

```python
gradebook = {
    "Alice": [90, 85, 92],
    "Bob": [78, 88, 95],
} # A dictionary where each value is a list of quiz scores

print(gradebook["Alice"]) # Print Alice's list of quiz scores

average_alice = sum(gradebook["Alice"]) / len(gradebook["Alice"]) # Compute Alice's average
print(f"Alice's average: {average_alice:.2f}") # Print the average, formatted to 2 decimals

for student_name, score_list in gradebook.items(): # Loop over each student and their scores
    avg = sum(score_list) / len(score_list)
    print(f"{student_name}'s average: {avg:.2f}")
```
