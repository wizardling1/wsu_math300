```python
def poly_growth(n, const, exp):
    return const * n**exp

x = np.logspace(2, 6, 5)
y = poly_growth(x, 1.0, 2)

fig, ax = plt.subplots(figsize=(4,4))
ax.loglog(x,y, "o-")

slope = np.polyfit(np.log10(x), np.log10(y), 1)[0]
print(f"Slope: {slope}")

fig.savefig("poly_growth.png")
plt.show()
```
