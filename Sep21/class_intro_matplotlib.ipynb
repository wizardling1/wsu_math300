```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x= np.linspace(-np.pi, np.pi, 10)
y =np.sin(x)

plt.plot(x,y, "--o", color="red")

plt.show()
```

```python
x= np.linspace(-np.pi, np.pi, 30)
y =np.sin(x**2)
y3= np.sin(x**3)
y2 = np.cos(x)

fig, ax = plt.subplots(nrows=1, ncols=2, figsize=(18,3), sharey= True)
ax[0].plot(x,y, "-", color="orangered",label =r"$\sin(x^2)$")
ax[0].plot(x,y3, "-s", color="green",label =r"$\sin(x^3)$")
ax[0].set_xlabel("x", fontsize=14)
ax[0].set_ylabel("y", fontsize=14)

ax[1].plot(x,y2, "--o", color="blue",label =r"$\cos(x)$")
ax[1].set_xlabel("x", fontsize=14)
ax[1].set_ylabel("y", fontsize=14)

plt.show()

```

```python
x= np.linspace(0, 10, 100)
y =np.linspace(0, 10, 100)

X, Y = np.meshgrid(x,y)

Z = np.sin(X) + np.cos(Y/2)

fig, ax = plt.subplots(figsize=(3,3))
c = ax.contourf(X,Y,Z, cmap="coolwarm", levels=20)

plt.colorbar(c, ax=ax)
plt.show()
```
